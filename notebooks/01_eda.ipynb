# Heart Disease UCI: EDA
Dataset: https://archive.ics.uci.edu/dataset/45/heart+disease (Cleveland subset, 303 rows). Run `python scripts/download_data.py` first.

```python
import sys; sys.path.insert(0, '../src')
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from heart.data import NUMERIC, CATEGORICAL, TARGET
df = pd.read_csv('../data/processed/heart_clean.csv')
sns.set_theme(style='whitegrid')
df.head()
```

## Missing values and class balance
Only `ca` (4) and `thal` (2) have missing values, so median/mode imputation is enough. Classes are close to balanced (about 54% / 46%), so accuracy is meaningful but we still report recall and ROC-AUC.

```python
print(df.isna().sum()[df.isna().sum() > 0])
df[TARGET].value_counts(normalize=True).plot.bar(title='Class balance')
```

## Numeric distributions by target

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 7))
for ax, c in zip(axes.flat, NUMERIC):
    sns.histplot(data=df, x=c, hue=TARGET, kde=True, ax=ax, element='step')
axes.flat[-1].axis('off')
```

## Correlation heatmap
Write your own observations here (e.g. which of `thalach`, `oldpeak`, `ca` correlate most with the target).

```python
plt.figure(figsize=(9, 7))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt='.2f', cmap='coolwarm', center=0)
```

## Categorical features vs target

```python
fig, axes = plt.subplots(2, 4, figsize=(18, 8))
for ax, c in zip(axes.flat, CATEGORICAL):
    sns.countplot(data=df, x=c, hue=TARGET, ax=ax)
```
